```python
# ============================================================
# ANALISIS COMPARATIVO RNA-PROTEINA
# CONDICIONES:
#   1. NLSK
#   2. KSLN
#   3. AAAA
#
# REGION ANALIZADA:
#   aa 179-182
#
# AUTOR:
#   Pipeline comparativo para 300 complejos
# ============================================================

# ============================================================
# INSTALAR DEPENDENCIAS
# ============================================================

!pip install -q MDAnalysis numpy pandas matplotlib seaborn networkx openpyxl

# ============================================================
# IMPORTAR LIBRERIAS
# ============================================================

import os
import tarfile
import tempfile
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import networkx as nx

from google.colab import files

import MDAnalysis as mda

from MDAnalysis.lib.distances import capped_distance

warnings.filterwarnings('ignore')

# ============================================================
# CONFIGURACION GENERAL
# ============================================================

REGION_INICIO = 179
REGION_FIN = 182

CUTOFF_DISTANCIA = 4.5

sns.set(style="whitegrid")

# ============================================================
# FUNCIONES AUXILIARES
# ============================================================

def obtener_elemento(atom_name):

    name_clean = atom_name.strip()

    if name_clean.startswith('C'):
        return 'C'

    elif name_clean.startswith('N'):
        return 'N'

    elif name_clean.startswith('O'):
        return 'O'

    elif name_clean.startswith('P'):
        return 'P'

    elif name_clean.startswith('S'):
        return 'S'

    else:
        return name_clean[0] if name_clean else 'X'


def es_anillo_aromatico(resname):

    return resname in ['PHE', 'TYR', 'TRP', 'HIS']


def es_base_nitrogenada(resname):

    return resname.strip().upper() in [
        'A', 'U', 'G', 'C',
        'DA', 'DT', 'DG', 'DC',
        'RA', 'RU', 'RG', 'RC'
    ]


def es_carga_positiva(resname):

    return resname in ['ARG', 'LYS', 'HIS']


def es_carga_negativa(resname):

    return resname in ['ASP', 'GLU']


def es_fosfato(atom_name):

    return atom_name.startswith('P') or atom_name.startswith('OP')


# ============================================================
# CLASIFICAR INTERACCIONES
# ============================================================

def clasificar_interaccion(atom_prot, atom_rna, dist):

    res_prot = atom_prot.resname
    res_rna = atom_rna.resname

    name_prot = atom_prot.name.strip()
    name_rna = atom_rna.name.strip()

    elem_prot = obtener_elemento(name_prot)
    elem_rna = obtener_elemento(name_rna)

    # Hydrogen bond
    if (
        (elem_prot in ['N', 'O']) and
        (elem_rna in ['N', 'O']) and
        dist < 3.2
    ):

        return 'Hydrogen_bond', 2.0

    # Aromatic stacking
    if (
        es_anillo_aromatico(res_prot) and
        es_base_nitrogenada(res_rna) and
        dist < 4.5
    ):

        return 'Stacking', 2.0

    # Salt bridge
    if (
        (
            es_carga_positiva(res_prot) or
            es_carga_negativa(res_prot)
        ) and
        es_fosfato(name_rna) and
        dist < 4.0
    ):

        return 'Salt_bridge', 2.0

    # Hydrophobic
    if (
        elem_prot == 'C' and
        elem_rna == 'C' and
        dist < 4.0
    ):

        return 'Hydrophobic', 1.0

    # Steric clash
    if dist < 2.5:

        return 'Steric_clash', -1.0

    # Van der Waals
    if dist < 4.5:

        return 'Van_der_Waals', 0.0

    return None, None


# ============================================================
# ANALISIS DE UN COMPLEJO
# ============================================================

def analizar_complejo(
    ruta_pdb,
    condicion
):

    try:

        u = mda.Universe(ruta_pdb)

        proteina = u.select_atoms("protein")
        rna = u.select_atoms("nucleic")

        if len(proteina) == 0 or len(rna) == 0:

            return pd.DataFrame()

        # ====================================================
        # REGION 179-182
        # ====================================================

        region = proteina.select_atoms(
            f"resid {REGION_INICIO}:{REGION_FIN}"
        )

        if len(region) == 0:

            return pd.DataFrame()

        # ====================================================
        # DISTANCIAS
        # ====================================================

        pairs, dists = capped_distance(
            region.positions,
            rna.positions,
            max_cutoff=CUTOFF_DISTANCIA,
            box=u.dimensions
        )

        registros = []

        for (idx_prot, idx_rna), d in zip(pairs, dists):

            atom_prot = region[idx_prot]
            atom_rna = rna[idx_rna]

            tipo, score = clasificar_interaccion(
                atom_prot,
                atom_rna,
                d
            )

            if tipo is not None:

                registros.append({

                    'condicion':
                        condicion,

                    'archivo':
                        os.path.basename(ruta_pdb),

                    'residuo_proteina':
                        f"{atom_prot.resname}{atom_prot.resid}",

                    'numero_residuo':
                        atom_prot.resid,

                    'atomo_proteina':
                        atom_prot.name,

                    'nucleotido_rna':
                        f"{atom_rna.resname}{atom_rna.resid}",

                    'base_rna':
                        atom_rna.resname.strip(),

                    'atomo_rna':
                        atom_rna.name,

                    'tipo_interaccion':
                        tipo,

                    'distancia_Ang':
                        round(d, 2),

                    'favorabilidad_score':
                        score,

                    'favorable':
                        'Sí' if score > 0 else (
                            'No' if score < 0 else 'Neutro'
                        )
                })

        return pd.DataFrame(registros)

    except Exception as e:

        print(f"Error en {ruta_pdb}: {e}")

        return pd.DataFrame()


# ============================================================
# EXTRAER ARCHIVOS
# ============================================================

def extraer_tar(tar_path):

    extract_dir = tempfile.mkdtemp()

    with tarfile.open(tar_path, 'r:gz') as tar:

        tar.extractall(extract_dir)

    return extract_dir


# ============================================================
# LISTAR PDBS
# ============================================================

def listar_pdbs(carpeta):

    lista_pdbs = []

    for root, dirs, files_in_dir in os.walk(carpeta):

        for f in files_in_dir:

            if f.lower().endswith(('.pdb', '.ent')):

                lista_pdbs.append(
                    os.path.join(root, f)
                )

    return lista_pdbs


# ============================================================
# ANALIZAR CONDICION
# ============================================================

def analizar_condicion(nombre_condicion):

    print("\n" + "="*70)
    print(f"📁 SUBIR ARCHIVO TAR.GZ PARA: {nombre_condicion}")
    print("="*70)

    uploaded = files.upload()

    tar_path = list(uploaded.keys())[0]

    print(f"\n📂 Extrayendo {nombre_condicion}...")

    extract_dir = extraer_tar(tar_path)

    lista_pdbs = listar_pdbs(extract_dir)

    print(f"✅ {len(lista_pdbs)} complejos encontrados")

    df_total = pd.DataFrame()

    for i, pdb_file in enumerate(lista_pdbs):

        print(
            f"Procesando {i+1}/{len(lista_pdbs)} "
            f"- {os.path.basename(pdb_file)}"
        )

        df_temp = analizar_complejo(
            pdb_file,
            nombre_condicion
        )

        df_total = pd.concat(
            [df_total, df_temp],
            ignore_index=True
        )

    return df_total, len(lista_pdbs)


# ============================================================
# ANALISIS DE LAS 3 CONDICIONES
# ============================================================

df_NLSK, n_NLSK = analizar_condicion("NLSK")

df_KSLN, n_KSLN = analizar_condicion("KSLN")

df_AAAA, n_AAAA = analizar_condicion("AAAA")


# ============================================================
# UNIFICAR DATASETS
# ============================================================

df_total = pd.concat(
    [
        df_NLSK,
        df_KSLN,
        df_AAAA
    ],
    ignore_index=True
)

print("\n" + "="*70)
print("✅ ANALISIS GLOBAL COMPLETADO")
print("="*70)

print(f"\n📊 Total interacciones:")
print(len(df_total))


# ============================================================
# GUARDAR CSV GENERAL
# ============================================================

df_total.to_csv(
    "interacciones_globales_300_complejos.csv",
    index=False
)

print("\n💾 CSV guardado")


# ============================================================
# RESUMEN ESTADISTICO
# ============================================================

resumen = []

for condicion, n_complejos in zip(
    ['NLSK', 'KSLN', 'AAAA'],
    [n_NLSK, n_KSLN, n_AAAA]
):

    df_cond = df_total[
        df_total['condicion'] == condicion
    ]

    complejos_con_interaccion = (
        df_cond['archivo'].nunique()
    )

    resumen.append({

        'Condicion':
            condicion,

        'Complejos_analizados':
            n_complejos,

        'Complejos_con_interaccion':
            complejos_con_interaccion,

        'Porcentaje_con_interaccion':
            round(
                complejos_con_interaccion /
                n_complejos * 100,
                2
            ),

        'Total_interacciones':
            len(df_cond),

        'Distancia_promedio':
            round(
                df_cond['distancia_Ang'].mean(),
                2
            ),

        'Favorabilidad_promedio':
            round(
                df_cond['favorabilidad_score'].mean(),
                2
            ),

        'Interaccion_mas_comun':
            (
                df_cond['tipo_interaccion']
                .mode()[0]
                if len(df_cond) > 0
                else 'NA'
            )
    })

df_resumen = pd.DataFrame(resumen)

print("\n" + "="*70)
print("📊 RESUMEN COMPARATIVO")
print("="*70)

print(df_resumen)

df_resumen.to_csv(
    "resumen_comparativo.csv",
    index=False
)


# ============================================================
# INTERPRETACION AUTOMATICA
# ============================================================

print("\n" + "="*70)
print("🧠 INTERPRETACION AUTOMATICA")
print("="*70)

total_NLSK = len(df_NLSK)
total_KSLN = len(df_KSLN)
total_AAAA = len(df_AAAA)

# ------------------------------------------------------------
# NLSK vs KSLN
# ------------------------------------------------------------

if total_KSLN < total_NLSK:

    print(
        "\n🔹 KSLN presenta menos interacciones "
        "que NLSK."
    )

    print(
        "   Esto sugiere que cambiar el orden "
        "de los residuos podría alterar "
        "la geometría local de reconocimiento."
    )

else:

    print(
        "\n🔹 KSLN mantiene o incrementa "
        "las interacciones respecto a NLSK."
    )

# ------------------------------------------------------------
# AAAA
# ------------------------------------------------------------

if total_AAAA < total_NLSK:

    print(
        "\n🔹 AAAA reduce las interacciones "
        "respecto a NLSK."
    )

    print(
        "   Esto sugiere que los residuos "
        "originales contribuyen "
        "al reconocimiento del RNA."
    )

# ------------------------------------------------------------
# DISTANCIAS
# ------------------------------------------------------------

dist_NLSK = df_NLSK['distancia_Ang'].mean()
dist_KSLN = df_KSLN['distancia_Ang'].mean()
dist_AAAA = df_AAAA['distancia_Ang'].mean()

print("\n🔹 Distancias promedio:")

print(f"   NLSK : {dist_NLSK:.2f} Å")
print(f"   KSLN : {dist_KSLN:.2f} Å")
print(f"   AAAA : {dist_AAAA:.2f} Å")


# ============================================================
# GRAFICA 1
# TOTAL INTERACCIONES
# ============================================================

plt.figure(figsize=(10,6))

sns.barplot(
    data=df_resumen,
    x='Condicion',
    y='Total_interacciones',
    palette='viridis'
)

plt.title(
    "Total de interacciones por condicion",
    fontsize=14
)

plt.ylabel("Numero de interacciones")

plt.savefig(
    "1_total_interacciones.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()


# ============================================================
# GRAFICA 2
# TIPOS DE INTERACCION
# ============================================================

plt.figure(figsize=(12,6))

df_tipos = pd.crosstab(
    df_total['condicion'],
    df_total['tipo_interaccion']
)

df_tipos.plot(
    kind='bar',
    stacked=True,
    figsize=(12,6)
)

plt.title(
    "Tipos de interaccion por condicion",
    fontsize=14
)

plt.ylabel("Numero de interacciones")

plt.xticks(rotation=0)

plt.savefig(
    "2_tipos_interaccion.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()


# ============================================================
# GRAFICA 3
# DISTRIBUCION DISTANCIAS
# ============================================================

plt.figure(figsize=(10,6))

sns.boxplot(
    data=df_total,
    x='condicion',
    y='distancia_Ang'
)

plt.title(
    "Distribucion de distancias",
    fontsize=14
)

plt.ylabel("Distancia (Å)")

plt.savefig(
    "3_boxplot_distancias.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()


# ============================================================
# GRAFICA 4
# HEATMAP
# ============================================================

heatmap_data = pd.crosstab(
    df_total['residuo_proteina'],
    df_total['condicion']
)

plt.figure(figsize=(8,6))

sns.heatmap(
    heatmap_data,
    annot=True,
    cmap='YlOrRd',
    fmt='d'
)

plt.title(
    "Frecuencia de interacciones por residuo",
    fontsize=14
)

plt.savefig(
    "4_heatmap_residuos.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()


# ============================================================
# GRAFICA 5
# NUCLEOTIDOS
# ============================================================

plt.figure(figsize=(12,6))

top_nucleotidos = (
    df_total
    .groupby(['condicion', 'base_rna'])
    .size()
    .reset_index(name='conteo')
)

sns.barplot(
    data=top_nucleotidos,
    x='base_rna',
    y='conteo',
    hue='condicion'
)

plt.title(
    "Nucleotidos RNA mas frecuentes",
    fontsize=14
)

plt.ylabel("Frecuencia")

plt.savefig(
    "5_nucleotidos_frecuentes.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()


# ============================================================
# EXPORTAR EXCEL MULTIHOJA
# ============================================================

with pd.ExcelWriter(
    'analisis_completo_300_complejos.xlsx'
) as writer:

    df_total.to_excel(
        writer,
        sheet_name='Interacciones',
        index=False
    )

    df_resumen.to_excel(
        writer,
        sheet_name='Resumen',
        index=False
    )

    df_NLSK.to_excel(
        writer,
        sheet_name='NLSK',
        index=False
    )

    df_KSLN.to_excel(
        writer,
        sheet_name='KSLN',
        index=False
    )

    df_AAAA.to_excel(
        writer,
        sheet_name='AAAA',
        index=False
    )

print("\n" + "="*70)
print("✅ ARCHIVOS GENERADOS")
print("="*70)

print("\n📁 CSV:")
print("   • interacciones_globales_300_complejos.csv")
print("   • resumen_comparativo.csv")

print("\n📁 Excel:")
print("   • analisis_completo_300_complejos.xlsx")

print("\n📁 Graficas:")
print("   • 1_total_interacciones.png")
print("   • 2_tipos_interaccion.png")
print("   • 3_boxplot_distancias.png")
print("   • 4_heatmap_residuos.png")
print("   • 5_nucleotidos_frecuentes.png")

print("\n🎉 Pipeline finalizado correctamente")
```
